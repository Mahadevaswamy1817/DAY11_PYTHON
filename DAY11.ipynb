```python
class Human:
  def eat(self):
    self.num_eyes = 2
    self.num_hands = 2
    print("i can walk")
  def work(self):
    print("i can work")

class Male(Human):
  def __init__(self, name):
    self.name = name

  def flirt(self):
    print("i am watching movies")

  def work(self):
    super().work() # using super function to call parent's work method
    print("i can code") # overriding method

s1 = Male("akash")
s1.eat()
s1.work()
s1.flirt()
print(s1.num_eyes)
```
